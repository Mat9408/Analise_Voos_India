```python
#Verificando se houve algum vôo com menos de 1h de duração
base[base.duration < 1.00]
```

```python
#Verificando se houve vôos noturnos para Dheli
base[(base.departure_time == 'Night') & (base.destination_city == 'Delhi')].head()
```

```python
#Utilizando o iloc para realizar o filtro acima
base.loc[(base.departure_time == "Night") & (base.destination_city == "Delhi")].head()
```

```python
#Utilizando argumentos lógicos
base.loc[(base.departure_time == "Night") | (base.destination_city == 'Delhi'), ['flight']].head()
```
